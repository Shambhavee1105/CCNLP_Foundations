```python
import nltk
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
```

```python
#stemming - the process of removing suffixes/prefixes from words to get a root form, which may not be a valid dictionary word
from nltk.stem import PorterStemmer
ps = PorterStemmer()
ps.stem("Running")
```

```python
ps.stem("Studies")
```

```python
#lemmatization - reduces a word to its lemma(dictionary form) using vocabulary and grammar(POS tagging)
from nltk.stem import WordNetLemmatizer
lm = WordNetLemmatizer()
lm.lemmatize("running", pos="v") #v = verb
```

```python
lm.lemmatize("better", pos="a") #a = adjective
```

```python
lm.lemmatize("studies", pos="n") #n = noun
```

```python
"""POS Tagging:

Nouns:
1. NN   – Noun, singular 
2. NNS  – Noun, plural
3. NNP  – Proper noun, singular
4. NNPS – Proper noun, plural

Verbs:
1. VB   – Verb, base form 
2. VBD  – Verb, past tense
3. VBG  – Verb, gerund/present participle
4. VBN  – Verb, past participle
5. VBP  – Verb, non-3rd person singular present 
6. VBZ  – Verb, 3rd person singular present 

Adjectives:
1. JJ   – Adjective
2. JJR  – Comparative adjective
3. JJS  – Superlative adjective

Adverbs:
1. RB   – Adverb
2. RBR  – Comparative adverb
3. RBS  – Superlative adverb
"""

from nltk import pos_tag
from nltk.tokenize import word_tokenize
text = "Students are learning Natural Language Processing"
words = word_tokenize(text)
pos_tag(words)
```

```python
ps.stem("Natural Language Processing is very interesting")
```

```python
lm.lemmatize("Natural Language Processing is very interesting")
```
