```python
import nltk
nltk.download("punkt")

translation_dict = {
    "hello": "hallo",
    "guys": "leute",
    "how": "wie",
    "are": "sind",
    "you": "du",
    "i": "ich",
    "am": "bin",
    "fine": "gut",
    "thank": "danke",
    "thanks": "danke",
    "what": "was",
    "is": "ist",
    "your": "dein",
    "name": "name",
    "my": "mein",
    "good": "gut",
    "morning": "morgen",
    "evening": "abend",
    "night": "nacht",
    "bye": "tschüss"
}

text = "hello guys! how are you?"

words = nltk.word_tokenize(text.lower())

translated_words = []

for word in words:
    translated_word = translation_dict.get(word, word)
    translated_words.append(translated_word)
    
translated_sentence = " ".join(translated_words)

print("Original Text: ")
print(text)
print("Translated text: ")
print(translated_sentence)
```
