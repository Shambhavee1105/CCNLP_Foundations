```python
from gtts import gTTS
import os

# Split text into languages
marathi = "आजच्या काळात NLP खूप वेगाने विकसित होत आहे आणि विविध उद्योगांमध्ये त्याचा मोठ्या प्रमाणावर वापर केला जात आहे."
gTTS(text=marathi, lang='mr').save("mr.mp3")
os.system("start mr.mp3")


hindi = "आज के समय में NLP तेजी से विकसित हो रहा है और कई उद्योगों में इसका उपयोग किया जा रहा है।"
gTTS(text=hindi, lang='hi').save("hi.mp3")
os.system("start hi.mp3")


english = "Today, NLP is rapidly evolving and is being used across industries such as healthcare, education, finance, and customer service to improve efficiency and user experience."
gTTS(text=english, lang='en').save("en.mp3")
os.system("start en.mp3")
```
