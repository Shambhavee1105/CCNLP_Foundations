```python
from ibm_watson import SpeechToTextV1, TextToSpeechV1
from ibm_cloud_sdk_core.authenticators import IAMAuthenticator
import pyaudio
import wave
import IPython.display as ipd

#credentials
#speech to text
STT_API_KEY = "Ewk-EKO5JH2sgXmMRGngIAiAGUSWxpMudZIBHDFEk6wa"
STT_URL = "https://api.au-syd.speech-to-text.watson.cloud.ibm.com/instances/938adfd4-7269-4447-b3ae-6269c7189b4a"

#text to speech
TTS_API_KEY = "jyPXBCc5q9mJrt-OMELJjDGnGPnroO9k8CUsVFotZ6N4"
TTS_URL = "https://api.au-syd.text-to-speech.watson.cloud.ibm.com/instances/505ff51d-3ce0-4062-844f-077d1223e2ac"


#record audio
def record_audio(filename="recorded.wav", duration=5):
    chunk = 1024
    format = pyaudio.paInt16
    channels = 1
    rate = 44100

    p = pyaudio.PyAudio()

    stream = p.open(format=format,
                    channels=channels,
                    rate=rate,
                    input=True,
                    frames_per_buffer=chunk)

    print("Recording... Speak now!")

    frames = []
    for _ in range(0, int(rate / chunk * duration)):
        data = stream.read(chunk)
        frames.append(data)

    print("Recording finished.")

    stream.stop_stream()
    stream.close()
    p.terminate()

    wf = wave.open(filename, 'wb')
    wf.setnchannels(channels)
    wf.setsampwidth(p.get_sample_size(format))
    wf.setframerate(rate)
    wf.writeframes(b''.join(frames))
    wf.close()

    return filename


#speech to text
def speech_to_text_mic():
    audio_file = record_audio()

    authenticator = IAMAuthenticator(STT_API_KEY)
    stt = SpeechToTextV1(authenticator=authenticator)
    stt.set_service_url(STT_URL)

    try:
        with open(audio_file, 'rb') as audio:
            response = stt.recognize(
                audio=audio,
                content_type='audio/wav',
                model='en-US_BroadbandModel'
            ).get_result()

        print("\nRecognized Text:")
        for res in response['results']:
            print(res['alternatives'][0]['transcript'])

    except Exception as e:
        print("Error:", e)


#text to speech
def text_to_speech():
    authenticator = IAMAuthenticator(TTS_API_KEY)
    tts = TextToSpeechV1(authenticator=authenticator)
    tts.set_service_url(TTS_URL)

    text = input("Enter text: ")

    try:
        with open('output.wav', 'wb') as audio_file:
            response = tts.synthesize(
                text=text,
                accept='audio/wav',
                voice='en-US_AllisonV3Voice'
            ).get_result()

            audio_file.write(response.content)

        print("Audio saved as output.wav")
        return 'output.wav'

    except Exception as e:
        print("Error:", e)
        return None


#play audio
def play_audio(file):
    return ipd.Audio(file)


#main  menu
def main():
    while True:
        print("\nIBM Watson Menu: ")
        print("1. Speech to Text (Mic)")
        print("2. Text to Speech")
        print("3. Exit")

        choice = input("Enter choice: ")

        if choice == '1':
            speech_to_text_mic()

        elif choice == '2':
            file = text_to_speech()
            if file:
                display(play_audio(file))

        elif choice == '3':
            print("Exiting program...")
            break

        else:
            print("Invalid choice. Try again.")


main()
```
