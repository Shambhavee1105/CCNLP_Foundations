```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

nltk.download('punkt')
nltk.download('stopwords')

documents = [
    "I love natural language processing",
    "Natural language processing is fun",
    "I love NLP"
]
```

```python
#remove stopwords
#Stop words are commonly used words in a language that are usually removed during text preprocessing because they carry little meaningful information for tasks like text classification or search.
stop_words = set(stopwords.words('english'))
all_words = []
for doc in documents:
    tokens = word_tokenize(doc.lower())
    filtered_tokens = [word for word in tokens if word.isalpha() and word not in stop_words]
    all_words.extend(filtered_tokens)
```

```python
#Bag of Words Representation
#Text representation technique in Natural Language Processing (NLP) where a document is converted into a set (or “bag”) of words, ignoring grammar and word order, but keeping word frequency.
bow = FreqDist(all_words)
print("Bag of Words Representation: \n")
for word, freq in bow.items():
    print(word, ":", freq)
```

```python
print("Vocabulary: \n")
print(list(bow.keys()))
```

```python
#N-grams are continuous sequences of n words used in Natural Language Processing (NLP) to capture context.
```

```python
#Trigrams (n = 3):
#A trigram is a sequence of three consecutive words that provides more context than unigrams and bigrams.
from nltk.util import ngrams
text = "This is a simple sentence for generating n-grams"
tokens = nltk.word_tokenize(text)
n=3
ngrams_output = list(ngrams(tokens,n))
print(ngrams_output)
```

```python
#Bigrams (n = 2):
#A bigram is a sequence of two consecutive words used to capture basic word relationships.
from nltk.util import ngrams
text = "This is a simple sentence for generating n-grams"
tokens = nltk.word_tokenize(text)
n=2
ngrams_output = list(ngrams(tokens,n))
print(ngrams_output)
```

```python
#Unigrams (n = 1):
#A unigram is a single word considered as an individual unit in text processing.
from nltk.util import ngrams
text = "This is a simple sentence for generating n-grams"
tokens = nltk.word_tokenize(text)
n=1
ngrams_output = list(ngrams(tokens,n))
print(ngrams_output)
```
