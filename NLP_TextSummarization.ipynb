```python
import nltk, heapq, math
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
nltk.download("punkt")
nltk.download("stopwords")

text = "Natural Language Processing (NLP) is a branch of Artificial Intelligence that focuses on enabling computers to understand, interpret, and generate human language. It combines linguistics, computer science, and machine learning to process text and speech data effectively. NLP techniques are used in applications such as chatbots, language translation, sentiment analysis, and speech recognition systems. By leveraging algorithms like tokenization, parsing, and deep learning models, NLP systems can extract meaningful insights from large volumes of unstructured data. As technology advances, NLP continues to improve human-computer interaction, making communication with machines more natural and efficient."
sentences = sent_tokenize(text)
stop_words = set(stopwords.words("english"))

word_freq = {}

for s in sentences:
    for w in word_tokenize(s.lower()):
        if w.isalnum() and w not in stop_words:
            word_freq[w] = word_freq.get(w,0)+1

tfidf = {}

for w in word_freq:
    idf = math.log(len(sentences)/(1+sum(1 for s in sentences if w in word_tokenize(s.lower()))))
    tfidf[w] = word_freq[w]*idf
    
sent_score = {}

for s in sentences:
    for w in word_tokenize(s.lower()):
        if w in tfidf:
            sent_score[s] = sent_score.get(s,0)+tfidf[w]
            
summary = " ".join(heapq.nlargest(2, sent_score, key = sent_score.get))

print("Original text: ")
print(text)
print("\nExtractive Summary: ")
print(summary)
```

```python
from sumy.summarizers.lsa import LsaSummarizer
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer

text = "Natural Language Processing (NLP) is a branch of Artificial Intelligence that focuses on enabling computers to understand, interpret, and generate human language. It combines linguistics, computer science, and machine learning to process text and speech data effectively. NLP techniques are used in applications such as chatbots, language translation, sentiment analysis, and speech recognition systems. By leveraging algorithms like tokenization, parsing, and deep learning models, NLP systems can extract meaningful insights from large volumes of unstructured data. As technology advances, NLP continues to improve human-computer interaction, making communication with machines more natural and efficient."

parser = PlaintextParser.from_string(text, Tokenizer("english"))
summarizer = LsaSummarizer()
summary = summarizer(parser.document, 2)

print("Original text: ")
print(text)
print("\nAbstractive Summary:")
for sent in summary:
    print(sent)
```
